```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
# apply
df["new_income"] = df["income"].apply(lambda x: "20%" if x > 50000 else "10%")
df
```

```python
# map
gender_map = {"Male": "M", "Female": "F", "Unknown": "U"}
df["gender"] = df["gender"].map(gender_map)
df
```

```python
# assign
df = df.assign(inc_income = df["income"] * 1.1)
df
```

```python
# replace
df["country"] = df["country"].replace("USA", "US")
df
```

```python
# rename
df = df.rename(columns={"income": "salary"})
```
