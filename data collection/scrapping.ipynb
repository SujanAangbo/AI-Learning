```python
import requests

URL = "https://www.scrapethissite.com/pages/simple/";
res = requests.get(URL)
```

```python
res
with open("scraped_data/countries.html", "w") as f: 
    f.write(res.text)

```

```python
from bs4 import BeautifulSoup

with open("scraped_data/countries.html", "r") as f: 
    read_content = f.read()
    
soup = BeautifulSoup(read_content, "lxml")
```

```python
country_tag = soup.find_all("h3");
country_details = []

for country in country_tag: 
    name = country.get_text(strip=True)
    population = country.find_next("div").select_one("span.country-population").get_text(strip=True)
    area = country.find_next("div").select_one("span.country-area").get_text(strip=True)
    country_details.append([name, population, area])
```

```python
import pandas as pd

st_data = pd.DataFrame(country_details, columns=["Country", "Population", "Area"]);
st_data
```

```python
# save collected data: 
st_data.to_csv("scraped_data/st_data.csv");
```
