```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
# fetching and storing multiple page data: 

page_count = 1
while True: 
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quote_data = soup.select("div.quote")

    if not quote_data: 
        print("Page finished")
        break

    with open(f"scraped_data/quote_{page_count}.html", "w") as f: 
        f.write(res.text)
        print(f"quote from page {page_count}")

    page_count += 1;
```

```python
# read data from the saved files:

quote_count = 1
frame = pd.DataFrame()

while True: 
    with open(f"scraped_data/quote_{quote_count}.html", "r") as f: 
    data = f.read()
    soup = BeautifulSoup(data, "lxml")

    quotes_container = soup.select("div.quote")

    formated_data = []

    for quote_container in quotes_container: 
        quote = quote_container.find_next("span").get_text(strip=True)

        author = quote_container.find_next("small", class_="author").get_text()

        keywords = quote_container.find_next("div").select_one("meta.keywords")["content"].split(",")

        formated_data.append([quote, author, keywords])

    frame.append(formated_data, columns=["quote", "author", "keywords"])
    

```
