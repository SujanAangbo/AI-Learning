```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.random.normal(50, 5, 100)

plt.boxplot(data, whis=1.0, tick_labels=["abc"])
plt.grid(True)
plt.show()
```

```python
data = np.arange(11)

d1 = data * 2
d2 = data ** 2
d3 = data ** 3

plt.subplot(2, 2, 1)
plt.plot(data, d1)

plt.subplot(2, 2, 2)
plt.plot(data, d2)

plt.subplot(2, 2, 3)
plt.plot(data, d3)
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
x = np.arange(len(days))

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],
    [18, 19, 17, 20, 21],
    [30, 32, 31, 33, 34],
    [25, 26, 24, 27, 28]
]
print(temperatures)

fig, ax = plt.subplots(2, 2,)

for i in range(2): 
    for j in range(2): 
        index = i * 2 + j
        ax[i][j].plot(x, temperatures[index], marker="o")
        ax[i][j].set_title("Temperature of " + cities[index])
        ax[i][j].set_xlabel("Days")
        ax[i][j].set_ylabel("Temperatures")
        ax[i][j].set_xticks(x, days)

fig.suptitle("Cities Temperature")
fig.tight_layout()
fig.show()
```
