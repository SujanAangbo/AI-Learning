```python
import numpy as np

oscar_revenue = [1005, 170, 427, 133, 232]
non_oscar_revenue = [378, 2788, 829, 185, 275]

year = np.array([2008, 2009, 2010, 2011, 2012])
width = 0.4

plt.bar(year - width / 2, oscar_revenue, width)
plt.bar(year + width / 2, non_oscar_revenue, width)
```
