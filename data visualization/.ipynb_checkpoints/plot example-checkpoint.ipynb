```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Plot monthly average temperatures (°C) for a city over one year. Add a title, axis labels, and a grid.

months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
temps = [3, 5, 10, 15, 21, 26, 29, 28, 22, 16, 9, 4]

x = np.arange(len(months))

plt.title("Months vs Temperature")

plt.plot(x, temps)
plt.xlabel("Months")
plt.ylabel("Temperature in (C)")

plt.xticks(x, months)
plt.grid()
plt.xlimit(min(x), max(x))
plt.show()
```

```python
# Plot two lines on the same chart — sales of Product A and Product B over 6 months. Add a legend.

months = ['Jan','Feb','Mar','Apr','May','Jun']
product_a = [120, 135, 150, 145, 160, 175]
product_b = [90, 100, 95, 110, 130, 125]

x = np.arange(len(months))
plt.plot(x, product_a, label="Product A")
plt.plot(x, product_b, label="Product B")

plt.legend()
plt.show()
```

```python
# Plot website daily visitors for two weeks. Use a dashed red line with circle markers. Annotate the peak day.

days = list(range(1, 15))
visitors = [200,220,195,310,280,260,190,215,230,205,340,290,270,180]

plt.plot(days, visitors, linestyle="--", color="red", marker="o")
print(days)

for i in range(len(days)): 
    if visitors[i] > 250: 
        plt.annotate(visitors[i], xy=(days[i], visitors[i]), xytext=(days[i] + 0.51, visitors[i]))

plt.show()
```

```python
# Plot three students' exam scores across 5 subjects. Customize line styles and colors for each student.

subjects = ['Math','Sci','Eng','His','Art']
alice = [85,90,78,92,88]
bob = [70,65,80,75,60]
carol = [92,88,95,85,91]

x = np.arange(len(subjects))

plt.plot(x, alice, label="alice", color="red", linestyle="--", marker="o")
plt.plot(x, bob, label="bob", color="blue", linestyle="-.", marker=">")
plt.plot(x, carol, label="carol", color="green", linestyle="-", marker="<")

plt.legend()

plt.show()
```

```python
# Create a vertical bar chart showing the number of books read by 5 friends in a year. Add value labels on top of each bar.

friends = ['Alice','Bob','Carol','David','Eve']
books = [24, 15, 30, 8, 20]

plt.bar(friends, books)

for i in range(len(friends)): 
    plt.text(friends[i], books[i] + 0.2, str(books[i]), ha="center")

plt.show()
```
